# Model

```python
import keras
import tensorflow as tf
import numpy as np

model = keras.Sequential([
    keras.layers.Input(shape=(2,)),
    keras.layers.Dense(16, 'relu'),
    keras.layers.Dense(2)
])

model.compile()

dummy_x=np.random.normal(size=(1,2))
dummy_y=model.predict(dummy_x)
```

# Zadania

## Export

Zapisać model jako `tmp.keras`

```python
model.save('tmp.keras')
model2 = keras.saving.load_model('tmp.keras')
y2 = model2.predict(dummy_x)
assert np.allclose(y2, dummy_y)
```

```python
model2.layers
```

## LiteRT

Wywołać eksportowany model za pomocą `LiteRT`

`pip install ai-edge-litert`

```python
converter = tf.lite.TFLiteConverter.from_keras_model(model)
tflite_model = converter.convert()
```

```python
with open("model.tflite", "wb") as f:
    f.write(tflite_model)
```
