```python
from bs4 import BeautifulSoup
from urllib.request import urlopen
```

```python
from selenium.webdriver import Firefox, Chrome, Edge
from selenium.webdriver import FirefoxOptions, ChromeOptions, EdgeOptions
```

```python
link = 'https://web-scraping.dev/product/13'

html = urlopen(link)

soup = BeautifulSoup(html.read(), 'html.parser')

imgs = soup.find_all("img", class_ = "product-img")

for img in imgs:
    link = img['src']

    file_name = link.split('/')[-1]

    open_link = urlopen(link)

    image = open_link.read()

    x = open(file_name, 'wb')
    x.write(image)
```

```python
driverFF = Firefox()

driverFF.get('https://web-scraping.dev/product/13')

print(driverFF.current_url)
print(driverFF.page_source)

driverFF.save_screenshot('ss.png')

driverFF.save_full_page_screenshot('ss.png')
```

```python
options = FirefoxOptions()
options.add_argument('--headless')

driverFF = Firefox(options=options)

driverFF.get('https://web-scraping.dev/product/13')
driverFF.save_full_page_screenshot('ss2.png')
driverFF.close()
```

```python
options = ChromeOptions()
options.add_argument('--headless=new')
options.add_argument('--disable-gpu')
driverChrome = Chrome(options=options)

driverChrome.close()
```
