Różnice między modelami Text-generation, Question-answering, Conversational:
- Text-generation - generuje tekst token po tokenie, jest używany m. in. do: pisania tekstów, generowania kodu, tworzenia dialogów.    
Przykładowo: dokończenie zdania „Litwo, ojczyzno moja...”

- Question-answering - odpowiada na pytania na podstawie podanego kontekstu. Model wyszukuje fragment tekstu zawierający odpowiedź.     
Przykładowo: Kontekst: Warszawa jest stolicą Polski. Pytanie do czatu: Jaka jest stolica Polski? Odpowiedź: Warszawa    

- Conversational - model zoptymalizowany do dialogu z użytkownikiem, który uwzględnia historię rozmowy i teoretycznie generuje bardziej naturalne odpowiedzi.   
Przykładowo: różne chatboty i asystenci AI. 

Model instruction-tuned vs Model chat-optimized:    
- Model instruction-tuned - to model, który po podstawowym treningu został dodatkowo wytrenowany na przykładach zawierających polecenia i poprawne odpowiedzi. W takich danych model widzi strukturę typu: instrukcja od użytkownika oraz odpowiedź, która powinna zostać wygenerowana. Dzięki temu uczy się, jak poprawnie reagować na polecenia w języku naturalnym, przykłądowo "napisz streszczenie", "wyjaśnij pojęcie", "przetłumacz tekst". Taki model lepiej rozumie intencję użytkownika i generuje bardziej uporządkowane oraz przydatne odpowiedzi.    

- Model chat-optimized - jest dodatkowo dostosowany specjalnie do prowadzenia rozmowy. W jego treningu wykorzystuje się dane w formie dialogów między użytkownikiem a asystentem. Model uczy się utrzymywać kontekst rozmowy, odpowiadać w naturalny sposób i reagować na kolejne pytania w tej samej konwersacji. Lepiej sprawdza się w chatbotach i asystentach AI. 


Base vs Instruct:
- Model Base to podstawowa wersja modelu językowego wytrenowana na dużej ilości tekstów. Jego głównym zadaniem jest przewidywanie kolejnych słów w zdaniu. Nie jest on specjalnie przystosowany do wykonywania poleceń użytkownika.

- Model Instruct to model, który został dodatkowo dotrenowant na zbiorach danych zawierających polecenia i odpowiedzi. Dzięki temu lepiej rozumie, czego oczekuje użytkownik. Jego odpowiedzi są zwykle bardziej jasne, logiczne i bezpośrednio związane z poleceniem. Modele typu Instruct są dlatego częściej wykorzystywane w chatbotach, asystentach AI

W praktyce oznacza to, że model Base jest bardziej 'surowy' i często używany jako punkt wyjścia do dalszego trenowania, natomiast model Instruct jest już przygotowany do bezpośredniego działania.

## Prompt engineering
Prompt engineering może wydawać się mało ciekawy, ale odpowiednio dobrane typy promptów są podstawą dobrze funkcjonującego systemu typu RAG, Chatbotów i Agentów AI.  
Zdefiniuj, podaj przykłady i potencjalne zastosowanie poniższych typów promptów:
- Zero-shot prompting
- Few-shot prompting
- Chain-of-thought prompting
- Meta-prompting/Instruction prompting
- RAG prompting


```python
# Miejsce na twoją odpowiedź
```

1. Zero-shot prompting  
`Definicja`: polega na zadaniu modelowi polecenia bez podawania jakichkolwiek przykładów. Model musi wykonać zadanie wyłącznie na podstawie wiedzy zdobytej podczas treningu.   
`Przykład:` Prompt - "Określ nastawienie tego zdania: Ten film był bardzo nudny", Odpowiedź - "Negatywny sentyment"     
`Zastosowanie:` szybkie zadania NLP, proste chatboty, analiza tekstu, generalnie zaletą jest prostota i szybkość, ale w trudniejszych zadaniach może działać gorzej.    

2. Few-shot prompting   
`Definicja:` polega na podaniu modelowi kilku przykładów rozwiązania zadania przed właściwym pytaniem. Dzięki temu model uczy się na podstawie wzorców w promptcie.     
`Przykład:` Prompt - Określ nastawienie zdania. 
Zdanie: "To było świetne"     
Nastawienie: pozytywny    
Zdanie: "To było fatalne"   
Nastawienie: negatywny      
Zdanie: "To było interesujące"      
Nastwaienie: ?      
Model dzięki przykładom łatwiej rozpoznaje wzorzec odpowiedzi.      
`Zastosowanie:` klasyfikacja tekstu, ekstrakcja informacji, formatowanie odpowiedzi, itp.   

3. Chain-of-thought prompting   
`Definicja:` polega na zachęceniu modelu do pokazania kolejnych kroków rozumowania prowadzących do odpowiedzi. Model generuje wtedy proces myślenia krok po kroku.      
`Przykład:` Prompt: Rozwiąż zadanie krok po kroku: "Jan ma 3 jabłka. Kupił jeszcze 5 jabłek, a potem zjadł 2, Ile jabłek mu zostało?" Odpowiedź modelu: "Jan miał 3 jabłka, Kupił 5 i ma 8, Zjadł 2 i zostało mu 6, zatem odpowiedź to 6"   
`Zastosowanie:` zadania matematyczne, logika, analiza problemów, systemy wspomagające decyzje. 

4. Meta-prompting / Instruction prompting   
`Definicja:` polega na jasnym określeniu instrukcji, roli modelu oraz oczekiwanego formatu odpowiedzi.  
`Przykład:` Prompt - "Jesteś ekspertem od marketingu, wyjaśnij w prosty sposób czym jest SEO, odpowiedź napisz w 3 punktach". Model otrzymuje: rolę, instrukcję, format odpowiedzi      
`Zastosowanie:` chatboty eksperckie, systemy edukacyjne, generowanie raportów

5. RAG prompting    
`Definicja:` polega na przekazaniu modelowi dodatkowego kontekstu pobranego z zewnętrznych źródeł (bazy danych, dokumentów, internetu). Model generuje odpowiedź na podstawie tego kontekstu.       
`Przykład:` Prompt: "Jakie jest największe miasto w Polsce po względem liczby ludności?" + dołączona tabela ze spisem miast i ich statystykami (również tej odnośnie populacji). Odpowiedź: "Warszawa - ok. 2 mln ludzi".       
`Zastosowanie:` systemy Q&A na dokumentach, chatboty firmowe, wyszukiwarki AI, analiza dokumentów i baz wiedzy.

## Parametry modelu
Przy użyciu parametru generation_config przeanalizuj wpływ wartości parametrów **temperature**, **top_p** i **top_k** na odpowiedzi modelu. Znajdź przykłady w których widać różnice w skrajnych wartościach parametrów. Odpowiedz słownie, ale również dodaj komórki z kodem dla znalezionych przykładów. 

```python
# Miejsce na twoją odpowiedź
```

```python
response = model.generate_content(
    "Ala ma kota, a co ma kot?",
    generation_config={
        "temperature": 0.1
    }
)

print(response.text)
```

```python
response = model.generate_content(
    "Ala ma kota, a co ma kot?",
    generation_config={
        "temperature": 1.2
    }
)

print(response.text)
```

Wydaje mi się, że dla temperatury znalazłem bardzo dobry przykład: Dla niskiej wartości (losowość jest mała), odpowiedź to Kot ma Alę, ale oczywiście też sierść... itp. Natomiast, gdy zwiększamy temperaturę do ponad 1, zwiększając losowość modelu, daje on trochę abstrakcyjną odpowiedź: Kot ma mruczenie... 

```python
response = model.generate_content(
    "Podaj mi 3 abstrakcyjne zastosowania AI w życiu codziennym - krótkie odpowiedzi",
    generation_config={
        "top_p": 0.01
    }
)

print(response.text)
```

```python
response = model.generate_content(
    "Podaj mi 3 abstrakcyjne zastosowania AI w życiu codziennym - krótkie odpowiedzi",
    generation_config={
        "top_p": 0.95
    }
)

print(response.text)
```

Dla TOP_P wykorzystałem taki prompt: "Podaj mi 3 abstrakcyjne zastosowania AI w życiu codziennym - krótkie odpowiedzi". Widać, że dla wyżej wartości model myśli bardziej abstrakcyjnie. "Antycypowanie potrzeb i personalizacja doświadczeń" brzmi prościej i mnie abstrakcyjnie niż "Antycypacyjna personalizacja doświadczeń" - model wykorzystał więcej słów, ale prostszych. 

```python
response = model.generate_content(
    "Podaj mi 3 tytuły książek o science fiction, które nie istnieją, ale mogłyby być interesujące",
    generation_config={
        "top_k": 1
    }
)

print(response.text)
```

```python
response = model.generate_content(
    "Podaj mi 3 tytuły książek o science fiction, które nie istnieją, ale mogłyby być interesujące",
    generation_config={
        "top_k": 100
    }
)

print(response.text)
```

Dla TOP_K ciekawy tytuł to Szepczące równoległe oraz mgławica myśląca. Są to bardzo abstracyjne odpowiedzi - nawet nie wiadomo co znaczą te tytuły. Tytuły dla niskich wartości są prostsze, przykładowo: algorytmiczna dżungla.

```python
import google.generativeai as genai

genai.configure(api_key="xxx")

model = genai.GenerativeModel(
    'gemini-2.5-flash',
    system_instruction="You are an AI expert explaining concepts to developers."
)

response = model.generate_content(
    "Explain what prompt engineering is."
)

print(response.text)
```

```python
import google.generativeai as genai

genai.configure(api_key="xxxx")

model = genai.GenerativeModel(
    'gemini-2.5-flash',
    system_instruction="Answer using table with 2 columns: 'Aspect' and 'Description', and 3 rows describing 3 different aspects of prompt engineering."
)

response = model.generate_content(
    "Explain what prompt engineering is."
)

print(response.text)
```

```python
import google.generativeai as genai

genai.configure(api_key="xxxx")

model = genai.GenerativeModel(
    'gemini-2.5-flash',
    system_instruction="Answer in exactly two sentences."
)

response = model.generate_content(
    "Explain what prompt engineering is."
)

print(response.text)
```

Dodanie promptu systemowego znacząco wpływa na sposób generowania odpowiedzi przez model. Ta sama treść pytania może prowadzić do różnych rezultatów w zależności od instrukcji systemowej.
Prompt systemowy może kontrolować styl odpowiedzi (np. tabela), poziom trudności (np. dla ekspertów), format odpowiedzi (np. max 2 sentencje), rolę modelu (np. jesteś ekspertem).
